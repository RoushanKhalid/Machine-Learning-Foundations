Variable, It's Types and Some Basic Operations

```python
a=10
b=20
b=15.999 # always pick the updated value stored in variable
complex_value_01=complex(2,3) # complex value: method 01
complex_value_02=2+3j # complex value: method 02
print(a)
print(b)
print("a+b") # text type (string)
print(a+b) # numeric type (int, float etc)
print(complex_value_01)
print(complex_value_02)
```

```python
first_name = "Roushan"
last_name = "Khalid"
full_name = first_name + " " + last_name  # Concatenated with space
print("Full Name:", full_name)
name_tuple = first_name, last_name       # Stored as a tuple
print("Tuple representation:", name_tuple)
print(first_name, last_name)             # Printed with separator
```

```python
student_age = 23
pi_constant = 3.14159
status_flag = "Active"
is_enrolled = True
complex_weight = complex(9, 7)
print("Age type:", type(student_age))
print("Pi type:", type(pi_constant))
print("Status type:", type(status_flag))
print("Enrolled type:", type(is_enrolled))
print("Complex weight type:", type(complex_weight))
```

Inputting, Explicit and Implicit Conversion

```python
researcher_name = "Khalid"
researcher_age = 23
hometown = "Dhaka"
profession = "ML Engineer"

print("Hi " + researcher_name + "!")
print("You are " + str(researcher_age) + " years old, from " + hometown + ", and work as a " + profession + " at Google.")
```

```python
val_str1 = "15"
val_str2 = "56"
concat_result = val_str1 + val_str2
print("String concatenation result:", concat_result)  # Output: 1556
```

```python
num1 = int("15")
num2 = int("56")
sum_result = num1 + num2
print("Integer addition sum:", sum_result)  # Output: 71
```

```python
x=5 # integer
y=3.14 # float
# Implicit casting: Integer is converted to float for addition
result=x+y
print(result) # Output: 8.14 (float)
print(type(result))

# Explicit casting: Float is converted to integer
result=x+int(y)
print(result) # Output: 3 (integer)
print(type(result))
```

String, Sub-String & Slicing

```python
sample_name = "Roushan Khalid Hasan"
print("Original text:", sample_name)
print("Index 8 char:", sample_name[8])
print("Slice [0:7]:", sample_name[0:7])
print("Slice [8:14]:", sample_name[8:14])
print("Slice [15:]:", sample_name[15:])
print("Slice [:-6]:", sample_name[:-6])
print("Complete slice [:]:", sample_name[:])
```
