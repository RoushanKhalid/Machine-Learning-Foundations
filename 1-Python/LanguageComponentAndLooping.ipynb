Looping

```python
step_count = 0
while step_count < 5:
    print("Training step iteration")
    step_count += 1
```

```python
limit = 5
accumulated_sum = 0
temp_num = limit
while temp_num > 0:
    accumulated_sum += temp_num
    temp_num -= 1
print(f"Sum of numbers from 1 to {limit} is: {accumulated_sum}")
```

```python
factorial_limit = 5
factorial_val = 1
counter = factorial_limit
while counter > 0:
    factorial_val *= counter
    counter -= 1
print(f"Factorial of {factorial_limit} is: {factorial_val}")
```

```python
max_limit = 9
product_accum = 1
while max_limit > 0:
    product_accum *= max_limit
    if product_accum > 100:
        break
    max_limit -= 1
print("Early termination product:", product_accum)
```

```python
num = 1
while num <= 10:
    if num == 7:
        num += 1
        continue
    print(num)
    num += 1
```

```python
sample_numbers = [6, 5, 2, 0]
total_sum = 0
for val in sample_numbers:
    if val == 0:
        break
    total_sum += val
print("Calculated loop sum:", total_sum)
```
