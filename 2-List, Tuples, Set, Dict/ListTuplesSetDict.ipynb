List Slicing

```python
slicing=["Apple", "Banana", "Cherry", "Mango", "Guava"]
print(slicing[0:2])  # Output: ['Apple', 'Banana']
print(slicing[2:])  # Output: ['Cherry', 'Mango', 'Guava']
print(slicing[-1])  # Output: Guava
print(slicing[-3:-1])  # Output: ['Cherry', 'Mango']
print(slicing[0:5])  # Output: ['Apple', 'Banana', 'Cherry', 'Mango', 'Guava']

# Extract every other element from index 0 to 5 (inclusive)
print(slicing[0:5:2])  # Output: ['Apple', 'Cherry', 'Guava']
print(slicing[-1:-2])  # Output: []

print(slicing[0:5:2]) # Repeat the previous slice operation  # Output: ['Apple', 'Cherry', 'Guava']
```

List Methods

```python
team = ["khalid", "hasan"]
team.append("farhan")
print("After append:", team)

senior_leads = ["tariq"]
senior_leads.extend(team)
print("After extend:", senior_leads)

senior_leads.insert(1, "rahim")
print("After insert:", senior_leads)

senior_leads.reverse()
print("After reverse:", senior_leads)

popped_member = senior_leads.pop(3)
print("Popped:", popped_member)
print("Remaining:", senior_leads)

senior_leads.remove("khalid")
print("After remove:", senior_leads)

senior_leads.sort()
print("Sorted:", senior_leads)

senior_leads.sort(reverse=True)
print("Reverse sorted:", senior_leads)

print("Index of tariq:", senior_leads.index("tariq"))
print("Count of tariq:", senior_leads.count("tariq"))

senior_leads.clear()
print("After clear:", senior_leads)
```

Write a program to make another list of duplicate elements from the following list : [1, 5, 6, 5, 1, 2, 3]

```python
my_list = [1, 5, 6, 5, 1, 2, 3]
duplicates = []

for i in range(len(my_list)):
  for j in range(i+1, len(my_list)):
    if my_list[i] == my_list[j]:
      duplicates.append(my_list[i])
      break

print(set(duplicates))  # Output: {1, 5}
```
